## Nettoyage

### Nettoyage BODACC

```python
bodacc_clean = bodacc.dropna(subset=["siren", "date_jugement"])

bodacc_clean = bodacc_clean.drop_duplicates()

bodacc_clean["date_parution"] = pd.to_datetime(bodacc_clean["date_parution"], errors="coerce")
bodacc_clean["date_jugement"] = pd.to_datetime(bodacc_clean["date_jugement"], errors="coerce")
```

```python
print(bodacc_clean["date_jugement"].min())
print(bodacc_clean["date_jugement"].max())
```

```python
bodacc_clean["date_jugement"].dt.year.value_counts().sort_index()
```

```python
bodacc_clean = bodacc_clean[bodacc_clean["date_jugement"].dt.year.between(2019, 2026)]
bodacc_clean = bodacc_clean.sort_values("date_jugement")
bodacc_clean = bodacc_clean.drop_duplicates(subset="siren", keep="first")
```

```python
print(bodacc.shape)
print(bodacc_clean.shape)
print(len(bodacc) - len(bodacc_clean))
bodacc_clean.head()
```

### Nettoyage SIRENE

```python
# garder les actives + les cessées qui ont une procédure BODACC
sirens_bodacc = bodacc_clean["siren"].unique()
mask_active = sirene["etatAdministratifUniteLegale"] == "A"
mask_cessees_bodacc = (sirene["etatAdministratifUniteLegale"] == "C") & (sirene["siren"].isin(sirens_bodacc))
sirene_clean = sirene[mask_active | mask_cessees_bodacc]

# colonnes utiles
COLS = [
    "siren",
    "dateCreationUniteLegale",
    "activitePrincipaleUniteLegale",
    "categorieJuridiqueUniteLegale",
    "trancheEffectifsUniteLegale",
    "categorieEntreprise",
]
sirene_clean = sirene_clean[COLS]
```

```python
print(sirene_clean["dateCreationUniteLegale"].dropna().min())
```

```python
# Conversion/suppression date de création
sirene_clean["dateCreationUniteLegale"] = pd.to_datetime(sirene_clean["dateCreationUniteLegale"], errors="coerce")
sirene_clean = sirene_clean[sirene_clean["dateCreationUniteLegale"].between("1807-01-01", "2026-08-01")]
```

```python
print(sirene.shape)
print(sirene_clean.shape)
print(len(sirene) - len(sirene_clean))
sirene_clean.head()
```

## Jointure BODACC + SIRENE

```python
# sirene + bodacc
dataset = sirene_clean.merge(
    bodacc_clean[["siren", "date_jugement", "nature_jugement"]],
    on="siren",
    how="left"
)

dataset["defaillance"] = dataset["date_jugement"].notna()

# âge de l'entreprise : à la date du jugement pour les défaillantes, sinon au 1er août 2026
date_ref = dataset["date_jugement"].fillna(pd.Timestamp("2026-08-01"))
dataset["age_entreprise"] = (date_ref - dataset["dateCreationUniteLegale"]).dt.days / 365.25

dataset.shape
```

```python
print(dataset["defaillance"].value_counts())
print(f"défaillances: {dataset['defaillance'].mean()}")
print(f"doublons: {dataset.duplicated().sum()}")
dataset.head()
```

```python
print("âges négatifs :", (dataset["age_entreprise"] < 0).sum())
dataset = dataset[dataset["age_entreprise"] >= 0]
```

```python
os.makedirs("../Data/processed", exist_ok=True)

dataset.to_parquet("../Data/processed/dataset_final.parquet", index=False)
print(dataset.shape)
```
